# 마크다운(Markdown)
   - 일반 텍스트 기반의 경량 마크업 언어로 웹표준이 아닌 언어이다.  
   마크업 언어란, 태그 등을 이용하여 문서나 데이터의 구조 등을 기술하며,  
   텍스트만으로 서식이 있는 문서들을 작성할 때 자주 사용된다. 또한 다른 마크업 언어들에 비해  
   문법이 쉽고 간단하여 HTML 등의 서식 문서들로 쉽게 변환되기 때문에 README 파일이나 온라인 게시물 등에 사용된다.

# 용어 
- [] : 대괄호 (브레이킷)  
- {} : 중괄호 (컬리 브레이킷)  
- () : 소괄호

## 줄바꿈 
- 띄어쓰기 (공백문자) 2개 적으면 줄바꿈 
- <br /> 태그를 사용하여 불바꿈 
- ex)파이썬을 귀도반 로섬이 좋아하는 코미디언의 이름은<br /> 몬티 파이썬이다

# 표현식
1. *내용* :글자 기울이기 (italic)
   - 파이썬은 퀴도반 로섬이 *좋아하는 코미디언의 이름*을 따서 만든 이름이다.
2. **내용**: 글자 굵게 (bold)
   -파이썬의 로고는 **그리스 로마 신화의 뱀**을 의미히여 생긴 로고이다.
4. ~~내용~~(~2 두 번): 취조선 (line thouth)
   - ~~c언어~~로 2년 걸린 프로젝트를 파이썬으로 한 달만에 마무리 했다.
5. '내용': 강조(back tick)
   -파이썬으 부모의언어는 'c언어 이다 즉, 파이썬은 c언어에서 파생되었다. 

#태그(tag)(element)
-여는 태그: <태그명>, 태그의 시작을 알리는 태그  
-닫는 태그: </태그명>, 태그의 종료를 알라는 태그  
-내용: <태그명>내용</태그명>, 태그가 포함하는 컨텐츠  
-ex) <sup>홍길동</sup>  
-<img src="주소" />  
-<img style="width:200px;"src="주소" /> :이미지 크기 조절
